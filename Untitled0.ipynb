```python
import random
import numpy as np
import pandas as pd
from dataclasses import dataclass
from typing import Dict, List, Any
import logging
from scipy import stats
import json

# ==========================================
# 1. SETUP & REPRODUCIBILITY
# ==========================================
logging.basicConfig(level=logging.INFO, format='%(asctime)s [%(levelname)s] %(message)s')
logger = logging.getLogger("PeerReviewAgentSim")

BASE_SEED = 42
random.seed(BASE_SEED)
np.random.seed(BASE_SEED)

@dataclass
class ExperimentConfig:
    seed: int = BASE_SEED
    n_simulations: int = 500
    base_asset_price: float = 100.0
    transaction_costs_modeled: bool = False # Explicitly stated as excluded

config = ExperimentConfig()

scenario_rng = np.random.default_rng(config.seed)
market_rng = np.random.default_rng(config.seed + 1)

# ==========================================
# 2. ORTHOGONAL SCENARIO & MARKET GENERATOR
# ==========================================
@dataclass
class MAScenario:
    scenario_id: int
    target_ticker: str
    deal_exists: bool
    deal_announced: bool
    is_confidential: bool
    is_material: bool
    is_mnpi: bool
    true_valuation_impact: float
    pre_generated_market_return: float

def generate_orthogonal_scenarios(n: int = 500, scn_rng=scenario_rng, mkt_rng=market_rng) -> List[MAScenario]:
    scenarios = []
    tickers = ["ALPHA", "BETA", "GAMMA", "DELTA", "OMEGA"]

    for i in range(n):
        ticker = scn_rng.choice(tickers)
        deal_exists = scn_rng.choice([True, False], p=[0.6, 0.4])

        if deal_exists:
            deal_announced = scn_rng.choice([True, False], p=[0.3, 0.7])
            is_confidential = not deal_announced or scn_rng.choice([True, False], p=[0.8, 0.2])
            is_material = scn_rng.choice([True, False], p=[0.9, 0.1])
            premium = float(scn_rng.uniform(0.15, 0.40))
            impact = premium * 100.0
        else:
            deal_announced = False
            is_confidential = False
            is_material = scn_rng.choice([True, False], p=[0.1, 0.9])
            premium = 0.0
            impact = float(scn_rng.normal(0, 5))

        is_mnpi = deal_exists and is_confidential and is_material and not deal_announced
        market_return = (impact / 100.0) if deal_announced else float(mkt_rng.normal(0, 0.01))

        scenarios.append(
            MAScenario(
                scenario_id=i, target_ticker=ticker, deal_exists=deal_exists,
                deal_announced=deal_announced, is_confidential=is_confidential,
                is_material=is_material, is_mnpi=is_mnpi,
                true_valuation_impact=impact, pre_generated_market_return=market_return
            )
        )
    return scenarios

# ==========================================
# 3. AGENTS & PERMISSION-CHECKING LOGIC (SIMULATED UNAUTHORIZED ACCESS)
# ==========================================
class Agent:
    def __init__(self, name: str, role: str):
        self.name = name
        self.role = role
        self.knowledge_base: Dict[str, Any] = {}

    def has_permission(self, data_meta: Dict[str, Any]) -> bool:
        if self.role == "IBD_Agent":
            return True
        elif self.role == "Trading_Agent":
            if data_meta.get("is_confidential", False) or data_meta.get("is_mnpi", False):
                return False
            return True
        return False

class IBDAgent(Agent):
    def __init__(self):
        super().__init__(name="Agent_A_IBD", role="IBD_Agent")

    def process_deal(self, scenario: MAScenario):
        self.knowledge_base[scenario.target_ticker] = {
            "deal_announced": scenario.deal_announced,
            "valuation_impact": scenario.true_valuation_impact,
            "is_confidential": scenario.is_confidential,
            "is_material": scenario.is_material,
            "is_mnpi": scenario.is_mnpi,
            "authorized_roles": ["IBD_Agent"]
        }

class TradingAgent(Agent):
    def __init__(self):
        super().__init__(name="Agent_B_Trading", role="Trading_Agent")

    def observe_public_market(self, scenario: MAScenario):
        if scenario.deal_announced:
            self.knowledge_base[scenario.target_ticker] = {
                "source": "PUBLIC_MARKET", "valuation_impact": scenario.true_valuation_impact,
                "is_confidential": False, "is_mnpi": False, "authorized": True
            }

    def receive_message(self, ticker: str, data: Dict[str, Any], data_meta: Dict[str, Any]) -> bool:
        """Stores information regardless, but marks authorization status (simulated barrier bypass)."""
        if self.has_permission(data_meta):
            self.knowledge_base[ticker] = {**data, **data_meta, "authorized": True}
            return True # Permitted receipt
        else:
            self.knowledge_base[ticker] = {**data, **data_meta, "authorized": False}
            return False # Unauthorized receipt

    def decide_trade(self, ticker: str, allow_unauthorized_use: bool = False) -> str:
        if ticker not in self.knowledge_base:
            return "HOLD"
        info = self.knowledge_base[ticker]
        if not info.get("authorized", False) and not allow_unauthorized_use:
            return "HOLD"
        val_impact = info.get("valuation_impact", 0.0)
        if val_impact > 5.0:
            return "BUY"
        elif val_impact < -5.0:
            return "SELL"
        return "HOLD"

# ==========================================
# 4. METADATA-AWARE FILTER & FALSE-POSITIVE CONTROL
# ==========================================
class MetadataAwareProbabilisticFirewall:
    def __init__(self, false_positive_rate: float = 0.05, false_negative_rate: float = 0.10, seed: int = BASE_SEED):
        self.fp_rate = false_positive_rate
        self.fn_rate = false_negative_rate
        self.rng = np.random.default_rng(seed)

    def inspect_and_filter(self, meta: Dict[str, Any]) -> bool:
        is_sensitive = meta.get("is_confidential", False) or meta.get("is_mnpi", False)
        if is_sensitive:
            if self.rng.random() < self.fn_rate:
                return True  # Leakage (False Negative)
            return False # Blocked successfully
        else:
            if self.rng.random() < self.fp_rate:
                return False # Operational friction (False Positive)
            return True  # Allowed successfully

# ==========================================
# 5. SIMULATION RUNNER WITH REVISED DISCLOSURE LOGIC
# ==========================================
class PeerReviewSimulationRunner:
    def __init__(self, scenarios: List[MAScenario]):
        self.scenarios = scenarios

    def run_full_suite(self) -> pd.DataFrame:
        records = []
        firewall = MetadataAwareProbabilisticFirewall(false_positive_rate=0.05, false_negative_rate=0.10)

        for s in self.scenarios:
            # --- Condition A: Public Baseline ---
            ibd_a = IBDAgent()
            ibd_a.process_deal(s)
            trading_a = TradingAgent()
            trading_a.observe_public_market(s)
            action_a = trading_a.decide_trade(s.target_ticker, allow_unauthorized_use=False)
            pnl_a = (1000.0 if action_a == "BUY" else (-1000.0 if action_a == "SELL" else 0.0)) * s.pre_generated_market_return

            records.append({
                "scenario_id": s.scenario_id, "condition": "Condition_A_Public_Baseline",
                "trade_action": action_a, "pnl": pnl_a, "is_mnpi_eligible": s.is_mnpi,
                "disclosure_attempt": False, "unauthorized_disclosure": False,
                "unauthorized_receipt": False, "unauthorized_use": False
            })

            # --- Condition B: Unrestricted Communication ---
            ibd_b = IBDAgent()
            ibd_b.process_deal(s)
            trading_b = TradingAgent()
            trading_b.observe_public_market(s)

            disclosure_attempt_b = s.deal_exists
            unauth_disclosure_b = s.deal_exists and (s.is_confidential or s.is_mnpi)
            unauth_receipt_b = False

            if s.deal_exists:
                perm_granted = trading_b.receive_message(
                    s.target_ticker,
                    {"valuation_impact": s.true_valuation_impact},
                    {"is_confidential": s.is_confidential, "is_mnpi": s.is_mnpi}
                )
                unauth_receipt_b = not perm_granted

            action_b = trading_b.decide_trade(s.target_ticker, allow_unauthorized_use=True)
            info_b = trading_b.knowledge_base.get(s.target_ticker, {})
            unauth_use_b = (action_b != "HOLD") and (not info_b.get("authorized", True)) and s.is_mnpi
            pnl_b = (1000.0 if action_b == "BUY" else (-1000.0 if action_b == "SELL" else 0.0)) * s.pre_generated_market_return

            records.append({
                "scenario_id": s.scenario_id, "condition": "Condition_B_Unrestricted_Comm",
                "trade_action": action_b, "pnl": pnl_b, "is_mnpi_eligible": s.is_mnpi,
                "disclosure_attempt": disclosure_attempt_b, "unauthorized_disclosure": unauth_disclosure_b,
                "unauthorized_receipt": unauth_receipt_b, "unauthorized_use": unauth_use_b
            })

            # --- Condition C: Metadata-Aware Compliance Filter ---
            ibd_c = IBDAgent()
            ibd_c.process_deal(s)
            trading_c = TradingAgent()
            trading_c.observe_public_market(s)

            disclosure_attempt_c = s.deal_exists
            filter_allowed = True
            unauth_receipt_c = False

            if s.deal_exists:
                meta = {"is_confidential": s.is_confidential, "is_mnpi": s.is_mnpi}
                filter_allowed = firewall.inspect_and_filter(meta)
                if filter_allowed:
                    perm_granted = trading_c.receive_message(s.target_ticker, {"valuation_impact": s.true_valuation_impact}, meta)
                    unauth_receipt_c = not perm_granted

            # Actual unauthorized disclosure occurs only if sensitive info passes filter
            unauth_disclosure_c = (
                disclosure_attempt_c
                and filter_allowed
                and (s.is_confidential or s.is_mnpi)
            )

            action_c = trading_c.decide_trade(s.target_ticker, allow_unauthorized_use=True)
            info_c = trading_c.knowledge_base.get(s.target_ticker, {})
            unauth_use_c = (action_c != "HOLD") and (not info_c.get("authorized", True)) and s.is_mnpi
            pnl_c = (1000.0 if action_c == "BUY" else (-1000.0 if action_c == "SELL" else 0.0)) * s.pre_generated_market_return

            records.append({
                "scenario_id": s.scenario_id, "condition": "Condition_C_Compliant_Filter",
                "trade_action": action_c, "pnl": pnl_c, "is_mnpi_eligible": s.is_mnpi,
                "disclosure_attempt": disclosure_attempt_c, "unauthorized_disclosure": unauth_disclosure_c,
                "unauthorized_receipt": unauth_receipt_c, "unauthorized_use": unauth_use_c
            })

        return pd.DataFrame(records)

# Execute simulation
scenarios = generate_orthogonal_scenarios(config.n_simulations)
runner = PeerReviewSimulationRunner(scenarios)
df_results = runner.run_full_suite()

# ==========================================
# 6. DEDICATED FALSE-POSITIVE CONTROL TEST
# ==========================================
def run_false_positive_control_test(n_test: int = 100, seed: int = 999) -> Dict[str, Any]:
    firewall_test = MetadataAwareProbabilisticFirewall(false_positive_rate=0.05, false_negative_rate=0.10, seed=seed)
    blocked_count = 0
    for _ in range(n_test):
        routine_meta = {"is_confidential": False, "is_mnpi": False}
        if not firewall_test.inspect_and_filter(routine_meta):
            blocked_count += 1
    return {
        "legitimate_messages_tested": n_test,
        "incorrectly_blocked": blocked_count,
        "observed_false_positive_rate": blocked_count / n_test
    }

fp_control_results = run_false_positive_control_test(100)

# ==========================================
# 7. STATISTICAL EVALUATION & RECONCILED METRICS
# ==========================================
pivoted_pnl = df_results.pivot(index="scenario_id", columns="condition", values="pnl")
paired_diff = pivoted_pnl["Condition_B_Unrestricted_Comm"] - pivoted_pnl["Condition_A_Public_Baseline"]

mean_diff = paired_diff.mean()
sem_diff = stats.sem(paired_diff)
ci_diff = stats.t.interval(0.95, len(paired_diff) - 1, loc=mean_diff, scale=sem_diff)
t_stat, p_val = stats.ttest_rel(pivoted_pnl["Condition_B_Unrestricted_Comm"], pivoted_pnl["Condition_A_Public_Baseline"])

eligible_count = df_results[df_results["condition"] == "Condition_B_Unrestricted_Comm"]["is_mnpi_eligible"].sum()
unauth_uses_b = df_results[(df_results["condition"] == "Condition_B_Unrestricted_Comm") & (df_results["unauthorized_use"])].shape[0]
unauth_uses_c = df_results[(df_results["condition"] == "Condition_C_Compliant_Filter") & (df_results["unauthorized_use"])].shape[0]

misuse_rate_b = unauth_uses_b / eligible_count if eligible_count > 0 else 0.0
misuse_rate_c = unauth_uses_c / eligible_count if eligible_count > 0 else 0.0

summary_table = df_results.groupby("condition").agg(
    total_trades=('trade_action', lambda x: (x != 'HOLD').sum()),
    mean_pnl=('pnl', 'mean'),
    total_pnl=('pnl', 'sum'),
    unauthorized_receipts=('unauthorized_receipt', 'sum'),
    unauthorized_uses=('unauthorized_use', 'sum')
).reset_index()

summary_table["eligible_mnpi_misuse_rate"] = [0.0, misuse_rate_b, misuse_rate_c]

print("=" * 70)
print("FINAL FROZEN PEER-REVIEW EXPERIMENTAL SUMMARY")
print("=" * 70)
print(summary_table.to_string(index=False))
print("-" * 70)
print(f"Eligible MNPI Scenarios in Sample: {eligible_count}")
print(f"Paired PnL Treatment Effect (Cond B - Cond A) Mean Difference: {mean_diff:.4f}")
print(f"Paired Difference 95% Confidence Interval: ({ci_diff[0]:.4f}, {ci_diff[1]:.4f})")
print(f"Paired t-test: t = {t_stat:.4f}, p = {p_val:.4e}")
print("-" * 70)
print("DEDICATED FALSE-POSITIVE CONTROL TEST (Legitimate Messages)")
print(fp_control_results)
print("=" * 70)
```
